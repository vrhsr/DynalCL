## 1. Mount Drive, get the frozen snapshot and the pilot bundle, verify hashes

```python
import hashlib, os, zipfile, shutil
from google.colab import drive
drive.mount("/content/drive")
os.makedirs("/content/drive/MyDrive/dynaicl_stage06_phaseD_v2", exist_ok=True)
EXPECTED = {"snapshot": "da53b423586fd58fdcd0d37f5ffaebd2a810088b3e13ad15b321b7b948765c23", "bundle": "c13dbd06313ee015c9cc8c464511cc1818f78ccac1d5f714614b0f5041b1d3a9"}
MANIFEST_SHA = "31a177841943ed1293a5b050d0037011e61dd34a45384ae86ad4cdb6b98c07ac"
os.environ["MANIFEST_SHA"] = MANIFEST_SHA
def sha(p): return hashlib.sha256(open(p, "rb").read()).hexdigest()
for name in ("stage06_frozen_snapshot_v2.zip", "phase_d_pilot_bundle_v2.zip"):
    on_drive = f"/content/drive/MyDrive/dynaicl_stage06_phaseD_v2/" + name
    if os.path.exists(on_drive):
        shutil.copy(on_drive, "/content/" + name)            # a restarted runtime finds them on Drive
    else:
        from google.colab import files
        print("upload", name); files.upload()
        shutil.copy("/content/" + name, on_drive)
assert sha("/content/stage06_frozen_snapshot_v2.zip") == EXPECTED["snapshot"], "snapshot hash mismatch"
assert sha("/content/phase_d_pilot_bundle_v2.zip") == EXPECTED["bundle"], "bundle hash mismatch"
shutil.rmtree("/content/DynaICL", ignore_errors=True); shutil.rmtree("/content/phase_d", ignore_errors=True)
zipfile.ZipFile("/content/stage06_frozen_snapshot_v2.zip").extractall("/content/DynaICL")
zipfile.ZipFile("/content/phase_d_pilot_bundle_v2.zip").extractall("/content/phase_d")
os.chdir("/content/DynaICL")
assert sha("/content/DynaICL/docs/FREEZE_MANIFEST_STAGE06_v2.json") == MANIFEST_SHA, "the extracted frozen manifest does not hash to the expected value"
print("hashes OK; the extracted manifest equals MANIFEST_SHA")
```
